# AutoPrepML creator workflow

This notebook downloads OpenML Adult v2 into a temporary cache, processes chunks in parallel, records a local experiment manifest, and removes all generated files when complete. It never embeds or writes data into the repository.

```python
from importlib.util import module_from_spec, spec_from_file_location
from pathlib import Path
script_path = Path('creator_examples/01_openml_pipeline.py')
if not script_path.exists():
    script_path = Path('01_openml_pipeline.py')
spec = spec_from_file_location('creator_example', script_path)
creator_example = module_from_spec(spec)
spec.loader.exec_module(creator_example)
result = creator_example.run_creator_example(rows=300, seed=42)
result
```
